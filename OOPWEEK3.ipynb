```python
class DigitalWallet:
  def __init__(self,name,balance,pin):
    self.name=name
    self.__balance=balance
    self.__pin=pin
  def deposit(self,amount):
    if amount>0:
      self.__balance+=amount
      return "Deposited successfully"
    return "Invalid amount"

  def withdraw(self,amount,pin):
    if self.__pin==pin and amount > 0 and self.__balance>=amount:
      self.__balance-=amount
      return "Amount is deducted"
    return "Something is wrong "

  @property
  def balance(self):
    return self.__balance

  def change_pin(self,old_pin,new_pin):
    if old_pin == self.__pin and  1000 <new_pin <9999:
      self.__pin =new_pin
      return "New pin is updated"
    return "Something is wrong"

w1 = DigitalWallet("Arshad", 5000, 1234)

print(w1.balance)

w1.deposit(2000)
print(w1.balance)

w1.withdraw(1000, 1234)
print(w1.balance)

print(w1.withdraw(500, 9999))

print(w1.change_pin(1234, 5678))

print(w1.withdraw(500, 5678))
print(w1.balance)
```

### `Problem-1:` Class inheritence

Create a **Bus** child class that inherits from the Vehicle class. The default fare charge of any vehicle is seating capacity * 100. If Vehicle is Bus instance, we need to add an extra 10% on full fare as a maintenance charge. So total fare for bus instance will become the final amount = total fare + 10% of the total fare.

Note: The bus seating capacity is 50. so the final fare amount should be 5500. You need to override the fare() method of a Vehicle class in Bus class.

```python
class Vehicle:
  def __init__(self,seating_capacity):
    self.seating_capacity=seating_capacity
  def fare(self):
    return self.seating_capacity*100
class Bus(Vehicle):
  def fare(self):
    total_fare=super().fare()
    maintenance=total_fare*0.10
    return total_fare+maintenance

b1 = Bus(50)

print("Total fare:", b1.fare())
```

```python
class Product:
  def __init__(self,name,price,quantity):
    self.name=name
    self.price=price
    self.quantity=quantity

class Order:
  def __init__(self,p1,p2,p3):
    self.p1=p1
    self.p2=p2
    self.p3=p3

  def show_products(self):
    prod=[self.p1,self.p2,self.p3]
    for p in prod:
      print(f"{p.name} - {p.price} - {p.quantity}")

  def total_bill(self):
    prod=[self.p1,self.p2,self.p3]
    total=0
    for tb in prod:
      total+=tb.price * tb.quantity

    return total

  def most_expensive(self):
    prod=[self.p1,self.p2,self.p3]
    high=prod[0]
    for m in prod:
      if m.price > high.price:
        high=m

    print(f"MostExpensive product : {high.name} - {high.price}")


# Creating Product objects
p1 = Product("Laptop", 50000, 1)
p2 = Product("Mouse", 1000, 2)
p3 = Product("Keyboard", 2000, 1)

# Creating Order object
order = Order(p1, p2, p3)

# Calling methods
order.show_products()
order.total_bill()
order.most_expensive()
```

```python
class Employee:
  def __init__(self,name,basic_salary):
    self.name=name
    self.basic_salary=basic_salary

  def calculate_salary(self):
    return self.basic_salary

class Developer(Employee):
  def __init__(self,name,basic_salary,experience):
    super().__init__(name,basic_salary)
    self.experience=experience
  def calculate_salary(self):
    salary=super().calculate_salary()
    return salary+(salary*0.10)

class SeniorDeveloper(Developer):
  def __init__(self,name,basic_salary,experience,team_size,project_bonus):
    super().__init__(name,basic_salary,experience)
    self.team_size=team_size
    self.project_bonus=project_bonus

  def calculate_salary(self):
    salary=super().calculate_salary()
    return salary+self.project_bonus

s = SeniorDeveloper("Arshad", 80000, 3, 5, 15000)

print(s.calculate_salary())
```

```python
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def show_person(self):
    print(f"Name:{self.name}")
    print(f"Age:{self.age}")

class Student:
  def __init__(self,roll_no,marks):
    self.roll_no=roll_no
    self.marks=marks

  def show_student(self):
    print(f"Roll:{self.roll_no}")
    print(f"Marks:{self.marks}")

class CollegeStudent(Person,Student):
  def __init__(self,name,age,roll_no,marks,branch):
    Person.__init__(self,name,age)
    Student.__init__(self,roll_no,marks)
    self.branch=branch

  def show_details(self):
    self.show_person()
    self.show_student()
    print(f"Branch:{self.branch}")

  def result(self):
    if self.marks >=90 and self.marks<=100:
      print("Excellent")
    elif self.marks >=75 and self.marks <=89:
      print("Very Good")
    elif self.marks >=60 and self.marks <=74:
       print("Good")
    elif self.marks >=40 and self.marks <=59:
      print("Pass")

    else:
      print("Fail")
s = CollegeStudent("Arshad", 20, 101, 85, "CSE")


s.show_details()
s.result()
```
