```python
class Circle:
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return 3.14*self.radius*self.radius
  def circumference(self):
    return 2*3.14*self.radius
  def display(self):
    print(f'The Area of the circle is {self.area()} nd also the circumference is {self.circumference()}')


c=Circle(6)
c.display()
```

```python
class Employee:
  def __init__(self,name,basic_salary):
    self.name=name
    self.basic_salary=basic_salary

  def calculate_bonus(self):
    if self.basic_salary >= 50000:
      self.bonus=self.basic_salary * 0.2
      return self.bonus
    elif self.basic_salary >= 30000:
      self.bonus=self.basic_salary*0.1
      return self.bonus
    else:
      self.bonus=self.basic_salary*0.05
      return self.bonus
  def total_salary(self):
    ts=self.basic_salary + self.bonus
    return ts
  def display(self):
    print(f'The Employee {self.name} is with salary {self.basic_salary} is getting the bonus of {self.calculate_bonus()} and the end his total salary is {self.total_salary()}')


e=Employee("Arshad",80000)
e.display()
```

```python
class BankAccount:
  def __init__(self,acc_no,name,balance):
    self.acc_no=acc_no
    self.name=name
    self.balance=balance

  def deposit(self,amount):
    self.balance+=amount
    return f"Deposited {amount} into {self.name}'s Account"

  def withdraw(self,amount):
    if self.balance >= amount:
      self.balance-=amount
      return f"Withdrawn {amount} from the {self.name}'s Account"
    else:
      return "Insufficient Balance"
  def transferamount(self,other,amount):
      if self.balance >= amount:
        self.balance -=amount
        other.balance+=amount
        return f"Transfered {amount} from {self.name} to {other.name}"
  def display(self):
    print(f"Account Number:{self.acc_no}")
    print(f"Name:{self.name}")
    print(f"Balance:{self.balance}")



a1 = BankAccount(101, "Arshad", 10000)
a2 = BankAccount(102, "Rahul", 3000)

a1.display()
print(a1.deposit(1000))
print(a1.withdraw(500))
print(a1.transferamount(a2, 2000))

print()
a1.display()
a2.display()
```

```python
class Book:
  def __init__(self,title,author,price):
    self.title=title
    self.author=author
    self.price=price
    self.available=True


  def borrow(self):
    if self.available:
      self.available=False
      return "Book borrowed successfully"

    return "Book is not available"

  def return_book(self):
    if not self.available :
      self.available=True
      return "Book returned succesfully"
    return "Book is already available"

  def display(self):
    print(f"Title:{self.title}")
    print(f"Author:{self.author}")
    print(f"Price:{self.price}")
    print(f"Is the book is Available: {self.available}")



b1=Book("Life of PI","Arshad",679)

b1.display()
print(b1.borrow())
print(b1.borrow())
print(b1.return_book())
b1.display()
```

```python
class Student:
  def __init__(self,name,roll_no,marks):
    self.name=name
    self.roll_no=roll_no
    self.marks=marks

  def average(self):
    total=0
    for mark in self.marks:
      total+=mark
    return total / len(self.marks)

  def highest(self):
    max=0
    for mark in self.marks:
      if mark > max:
        max=mark
    return max

  def lowest(self):
    min=100
    for mark in self.marks:
      if mark < min:
        min=mark
    return min

  def grade(self):
    if self.average() >=90:
      return "A grade"
    elif 75 <= self.average() <=89:
      return "B grade"
    elif 60 <= self.average() <=74:
      return "C grade"
    elif 40 <= self.average() <=59:
      return "D grade"
    else:
      return "F grade"

  def display(self):
    print(f"Name:{self.name}")
    print(f"Roll number:{self.roll_no}")
    print(f"Average marks:{self.average()}")
    print(f"Highest marks: {self.highest()}")
    print(f"lowest marks: {self.lowest()}")
    print(f"Grade:{self.grade()}")



s1 = Student("Arshad", 101, [85, 92, 78, 90])
s1.display()
```

```python
class Player:
  def __init__(self,name,score):
    self.name=name
    self.score=score

  def display(self):
    print(f"Name:{self.name}")
    print(f"Score:{self.score}")

class Team:
  def __init__(self,team_name):
    self.team_name = team_name
    self.players=[]

  def add_player(self,player):
    self.players.append(player)

  def total_score(self):
    total=0
    for player in self.players:
      total+=player.score
    return total

  def top_player(self):
    top=self.players[0]
    for player in self.players:
      if player.score > top.score:
        top=player

    return top

  def display(self):
    print(f"Team: {self.team_name}")
    print("Players:")

    for player in self.players:
        print(f"{player.name} - {player.score}")

    print(f"Total Score: {self.total_score()}")

    top = self.top_player()
    print(f"Top Player: {top.name}")



  p1 = Player("Arshad", 80)
p2 = Player("Rahul", 70)
p3 = Player("Sukumar", 90)

team = Team("Avengers")

team.add_player(p1)
team.add_player(p2)
team.add_player(p3)

team.display()

```

```python
class Book:
  def __init__(self,title,author,price):
    self.title=title
    self.author=author
    self.price=price

  def display(self):
    print(f"Title:{self.title}")
    print(f"Author:{self.author}")
    print(f"Price:{self.price}")


class Library:
  def __init__(self,name):
    self.name=name
    self.books=[]

  def add_book(self,book):
    self.books.append(book)

  def total_value(self):
    total=0
    for book in self.books:
      total+=book.price
    return total


  def most_expensive(self):
    exp=self.books[0]
    for book in self.books:
      if book.price > exp.price:
        exp=book
    return exp

  def display(self):
    for book in self.books:
      print(f"{book.title} - {book.author} - {book.price}")
    print(f"Total value of the books:{self.total_value()}")
    exp=self.most_expensive()
    print(f"Most expensive Book : {exp.title}")


b1 = Book("Python Basics", "Arshad", 500)
b2 = Book("Data Science", "Rahul", 800)
b3 = Book("Machine Learning", "Sukumar", 1200)

library = Library("Central Library")

library.add_book(b1)
library.add_book(b2)
library.add_book(b3)

library.display()
```

```python
class Product:
  def __init__(self,name,price,quantity):
    self.name=name
    self.price=price
    self.quantity=quantity

  def display(self):
    print(f"Name:{self.name}")
    print(f"Price:{self.price}")
    print(f"Quantity:{self.quantity}")


class Store:
  def __init__(self,store_name):
    self.store_name=store_name
    self.products=[]

  def add_product(self,product):
    self.products.append(product)

  def total_value(self):
    total=0
    for product in self.products:
      total+=product.price*product.quantity
    return total

  def sell_product(self,name,quantity):
    for product in self.products:
      if product.name == name:
        if product.quantity >=quantity:
          product.quantity-=quantity
          print(f"{product.name} sold {quantity}")
          return

        else:
          print("Not Enough Stock")
          return

    print("product Not found")

  def display(self):
    for product in self.products:
      print(f"{product.name} - {product.price} - {product.quantity}")

    print(f"Total Inventory Value: {self.total_value()}")


p1 = Product("Laptop", 50000, 2)
p2 = Product("Mouse", 1000, 5)
p3 = Product("Keyboard", 2000, 3)

store = Store("Arshad Store")

store.add_product(p1)
store.add_product(p2)
store.add_product(p3)

store.sell_product("Laptop", 1)

store.display()
```
